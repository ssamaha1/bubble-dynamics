```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import time
import pandas as pd

from scipy.integrate import solve_ivp
from scipy.interpolate import interp1d
from tensorflow.keras import Model
from tensorflow.keras.layers import Dense, LSTM, GRU
```

```python
# generating the data set

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

def rayleigh_plesset_rhs(t, y, params):
    """
    Stable Rayleigh-Plesset ODE system.

    State:
        y[0] = R(t)
        y[1] = Rdot(t)

    Equation:
        R Rddot + (3/2) Rdot^2 =
        (1/rho)[PB(R) - Pinf(t) - 2 sigma/R - 4 mu Rdot/R]
    """

    R = y[0]
    Rdot = y[1]

    rho = params["rho"]
    sigma = params["sigma"]
    mu = params["mu"]
    P0 = params["P0"]
    Pv = params["Pv"]
    R0 = params["R0"]
    kappa = params["kappa"]
    Pa = params["Pa"]
    freq = params["freq"]

    # Safety floor to avoid singular behavior at R = 0.
    R_safe = max(R, 0.25 * R0)

    # Acoustic forcing.
    Pinf = P0 + Pa * np.sin(2.0 * np.pi * freq * t)

    # Polytropic gas pressure.
    PB = (
        (P0 - Pv + 2.0 * sigma / R0)
        * (R0 / R_safe) ** (3.0 * kappa)
        + Pv
    )

    pressure_term = (
        PB
        - Pinf
        - 2.0 * sigma / R_safe
        - 4.0 * mu * Rdot / R_safe
    )

    Rddot = (pressure_term / rho - 1.5 * Rdot**2) / R_safe

    # Numerical safety.
    if not np.isfinite(Rddot):
        Rddot = 0.0

    return [Rdot, Rddot]


def collapse_event(t, y, params):
    """
    Stop integration if the radius collapses too much.
    """
    R = y[0]
    R0 = params["R0"]
    return R - 0.25 * R0

collapse_event.terminal = True
collapse_event.direction = -1


def generate_single_sample(t, rng):
    """
    Generate one stable ultrasound-driven microbubble trajectory.
    """

    params = {}

    # Soft-tissue / water-like medium.
    params["rho"] = 993.0             # kg/m^3
    params["sigma"] = 0.071           # N/m
    params["mu"] = rng.uniform(0.001, 0.004)   # Pa*s
    params["P0"] = 101325.0           # Pa
    params["Pv"] = 2330.0             # Pa
    params["kappa"] = 1.4

    # Microbubble-scale radius.
    params["R0"] = rng.uniform(0.8e-6, 3.0e-6)

    # Conservative acoustic forcing to avoid violent collapse.
    params["Pa"] = rng.uniform(30e3, 250e3)

    # Biomedical ultrasound frequency range.
    params["freq"] = rng.uniform(0.5e6, 1.5e6)

    y0 = [params["R0"], 0.0]

    sol = solve_ivp(
        fun=lambda tt, yy: rayleigh_plesset_rhs(tt, yy, params),
        t_span=(t[0], t[-1]),
        y0=y0,
        t_eval=t,
        method="RK45",
        rtol=1e-5,
        atol=1e-10,
        max_step=2e-8,
        events=lambda tt, yy: collapse_event(tt, yy, params)
    )

    if not sol.success:
        return None

    # If event stopped early, reject sample.
    if sol.y.shape[1] != len(t):
        return None

    R = sol.y[0]

    if np.any(~np.isfinite(R)):
        return None

    if np.any(R <= 0):
        return None

    # Reject physically extreme excursions.
    if np.max(R) > 8.0 * params["R0"]:
        return None

    Pinf = params["P0"] + params["Pa"] * np.sin(
        2.0 * np.pi * params["freq"] * t
    )

    del_p = Pinf - params["P0"]

    return del_p, R, params


def generate_dataset(n_samples, n_time=300, seed=0):
    """
    Generate stable Rayleigh-Plesset dataset.

    Output keys:
        del_p : pressure forcing, shape [N, T]
        R     : radius response, shape [N, T]
        t     : time vector, shape [T]
    """

    rng = np.random.default_rng(seed)

    # Shorter window = much faster and more stable.
    t = np.linspace(0.0, 5e-5, n_time)

    del_p_list = []
    R_list = []
    params_list = []

    attempts = 0
    max_attempts = 50 * n_samples

    while len(R_list) < n_samples and attempts < max_attempts:
        attempts += 1

        sample = generate_single_sample(t, rng)

        if sample is None:
            continue

        del_p, R, params = sample

        del_p_list.append(del_p)
        R_list.append(R)
        params_list.append(params)

        if len(R_list) % 25 == 0:
            print(f"Generated {len(R_list)} / {n_samples} samples")

    if len(R_list) < n_samples:
        print(f"Warning: only generated {len(R_list)} stable samples.")

    return {
        "del_p": np.array(del_p_list),
        "R": np.array(R_list),
        "t": t,
        "params": params_list
    }


# Smaller dataset first. You can increase later if it runs well.
train_data = generate_dataset(
    n_samples=200,
    n_time=300,
    seed=42
)

test_data = generate_dataset(
    n_samples=50,
    n_time=300,
    seed=123
)

np.savez(
    "/content/res_train.npz",
    del_p=train_data["del_p"],
    R=train_data["R"],
    t=train_data["t"]
)

np.savez(
    "/content/res_test.npz",
    del_p=test_data["del_p"],
    R=test_data["R"],
    t=test_data["t"]
)

print("train del_p:", train_data["del_p"].shape)
print("train R:", train_data["R"].shape)
print("test del_p:", test_data["del_p"].shape)
print("test R:", test_data["R"].shape)
print("time:", train_data["t"].shape)
```

```python
def preprocess(dataset, m, npoints_output, is_test=False):
    p = dataset["del_p"]
    t = dataset["t"]
    r = dataset["R"]

    P = interp1d(t, p, kind="cubic", axis=1)
    R_interp = interp1d(t, r, kind="cubic", axis=1)

    t_min = 0.0
    t_max = 5e-5

    t_branch = np.linspace(t_min, t_max, m)
    t_output = np.linspace(t_min, t_max, npoints_output)

    X_func = P(t_branch)
    y = R_interp(t_output)

    if is_test:
        X_func = X_func[:50]
        y = y[:50]

    return X_func, y, t_branch, t_output


def normalize(X_func, y, Par):
    X_func = (X_func - Par["p_mean"]) / Par["p_std"]
    y = (y - Par["r_mean"]) / Par["r_std"]

    return X_func.astype(np.float32), y.astype(np.float32)


data_train = np.load("/content/res_train.npz")
data_test = np.load("/content/res_test.npz")

m = 20
npoints_output = 20

X_train, y_train, t_branch, t_output = preprocess(
    data_train,
    m,
    npoints_output
)

X_test, y_test, _, _ = preprocess(
    data_test,
    m,
    npoints_output,
    is_test=True
)

Par = {}
Par["p_mean"] = np.mean(X_train)
Par["p_std"] = np.std(X_train)
Par["r_mean"] = np.mean(y_train)
Par["r_std"] = np.std(y_train)
Par["m"] = m
Par["npoints_output"] = npoints_output
Par["address"] = "/content/"

X_train, y_train = normalize(X_train, y_train, Par)
X_test, y_test = normalize(X_test, y_test, Par)

X_train_seq = X_train.reshape(-1, m, 1)
X_test_seq = X_test.reshape(-1, m, 1)
y_train_seq = y_train.reshape(-1, m, 1)
y_test_seq = y_test.reshape(-1, m, 1)

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_train_seq:", X_train_seq.shape)
print("y_train_seq:", y_train_seq.shape)
```

```python
def relative_l2_error(y_pred, y_true):
    return np.linalg.norm(y_pred.reshape(-1, 1) - y_true.reshape(-1, 1)) / np.linalg.norm(y_true.reshape(-1, 1))


def plot_convergence(history, title, filename):
    plt.figure(figsize=(7, 5))
    plt.plot(history["epoch"], history["train_loss"], label="train")
    plt.plot(history["epoch"], history["val_loss"], label="validation")
    plt.yscale("log")
    plt.xlabel("Epoch")
    plt.ylabel("MSE")
    plt.title(title)
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.savefig(filename, dpi=300)
    plt.show()


def plot_prediction(t, y_true, y_pred, title, filename, idx=0):
    plt.figure(figsize=(7, 5))
    plt.plot(t, y_true[idx], label="true")
    plt.plot(t, y_pred[idx], "--", label="predicted")
    plt.xlabel("Time (s)")
    plt.ylabel("Normalized radius")
    plt.title(title)
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.savefig(filename, dpi=300)
    plt.show()
```

```python
class LSTM_Model(Model):
    def __init__(self, Par):
        super().__init__()

        self.Par = Par
        self.optimizer = tf.keras.optimizers.Adam(learning_rate=1e-4)

        self.model = tf.keras.Sequential([
            LSTM(50),
            Dense(Par["m"])
        ])

    def call(self, X):
        pred = self.model(X)
        return tf.reshape(pred, [-1, self.Par["m"], 1])


@tf.function
def train_step_sequence(model, X, y):
    with tf.GradientTape() as tape:
        y_hat = model(X)
        loss = tf.reduce_mean(tf.square(y_hat - y))

    gradients = tape.gradient(loss, model.trainable_variables)
    model.optimizer.apply_gradients(zip(gradients, model.trainable_variables))

    return loss


def train_sequence_model(model, name, X_train, y_train, X_test, y_test, n_epochs=3000, batch_size=150):
    history = {"epoch": [], "train_loss": [], "val_loss": []}
    begin_time = time.time()

    for epoch in range(n_epochs + 1):
        for end in np.arange(batch_size, X_train.shape[0] + 1, batch_size):
            start = end - batch_size
            loss = train_step_sequence(model, X_train[start:end], y_train[start:end])

        if epoch % 500 == 0:
            y_hat = model(X_test).numpy()
            val_loss = np.mean((y_hat - y_test) ** 2)

            print(
                f"{name} epoch {epoch}: "
                f"train={loss.numpy():.3e}, "
                f"val={val_loss:.3e}, "
                f"time={int(time.time()-begin_time)}s"
            )

            history["epoch"].append(epoch)
            history["train_loss"].append(loss.numpy())
            history["val_loss"].append(val_loss)

    return history


lstm_model = LSTM_Model(Par)

lstm_history = train_sequence_model(
    lstm_model,
    "LSTM",
    X_train_seq,
    y_train_seq,
    X_test_seq,
    y_test_seq
)

lstm_pred = lstm_model(X_test_seq).numpy()

lstm_err = relative_l2_error(lstm_pred, y_test_seq)

print("LSTM relative L2 error:", lstm_err)

plot_convergence(
    lstm_history,
    "LSTM convergence",
    "/content/lstm_convergence.png"
)

plot_prediction(
    t_output,
    y_test,
    lstm_pred[:, :, 0],
    "LSTM prediction",
    "/content/lstm_prediction.png",
    idx=0
)
```

```python
# ============================================================
# GRU MODEL — CORRECTED VERSION
# ============================================================

class GRU_Model(tf.keras.Model):
    def __init__(self, Par):
        super().__init__()

        self.Par = Par

        self.optimizer = tf.keras.optimizers.Adam(
            learning_rate=1e-4
        )

        self.gru = tf.keras.layers.GRU(
            50,
            return_sequences=False
        )

        self.dense = tf.keras.layers.Dense(
            Par["m"]
        )

    def call(self, X, training=False):
        X = tf.convert_to_tensor(
            X,
            dtype=tf.float32
        )

        h = self.gru(
            X,
            training=training
        )

        pred = self.dense(
            h
        )

        pred = tf.reshape(
            pred,
            [-1, self.Par["m"], 1]
        )

        return pred


def train_step_sequence(model, X, y):
    X = tf.convert_to_tensor(
        X,
        dtype=tf.float32
    )

    y = tf.convert_to_tensor(
        y,
        dtype=tf.float32
    )

    with tf.GradientTape() as tape:
        y_hat = model(
            X,
            training=True
        )

        loss = tf.reduce_mean(
            tf.square(y_hat - y)
        )

    gradients = tape.gradient(
        loss,
        model.trainable_variables
    )

    model.optimizer.apply_gradients(
        zip(
            gradients,
            model.trainable_variables
        )
    )

    return loss


def train_sequence_model(
    model,
    name,
    X_train,
    y_train,
    X_test,
    y_test,
    n_epochs=3000,
    batch_size=150
):
    history = {
        "epoch": [],
        "train_loss": [],
        "val_loss": []
    }

    begin_time = time.time()

    for epoch in range(n_epochs + 1):

        for end in np.arange(
            batch_size,
            X_train.shape[0] + 1,
            batch_size
        ):
            start = end - batch_size

            loss = train_step_sequence(
                model,
                X_train[start:end],
                y_train[start:end]
            )

        if epoch % 500 == 0:
            y_hat = model(
                X_test,
                training=False
            ).numpy()

            val_loss = np.mean(
                (y_hat - y_test) ** 2
            )

            print(
                f"{name} epoch {epoch}: "
                f"train={loss.numpy():.3e}, "
                f"val={val_loss:.3e}, "
                f"time={int(time.time() - begin_time)}s"
            )

            history["epoch"].append(epoch)
            history["train_loss"].append(loss.numpy())
            history["val_loss"].append(val_loss)

    return history


gru_model = GRU_Model(
    Par
)

gru_history = train_sequence_model(
    gru_model,
    "GRU",
    X_train_seq,
    y_train_seq,
    X_test_seq,
    y_test_seq
)

gru_pred = gru_model(
    X_test_seq,
    training=False
).numpy()

gru_err = relative_l2_error(
    gru_pred,
    y_test_seq
)

print(
    "GRU relative L2 error:",
    gru_err
)

plot_convergence(
    gru_history,
    "GRU convergence",
    "/content/gru_convergence.png"
)

plot_prediction(
    t_output,
    y_test,
    gru_pred[:, :, 0],
    "GRU prediction",
    "/content/gru_prediction.png",
    idx=0
)
```

```python
class Seq2Seq_GRU_Model(Model):
    def __init__(self, Par):
        super().__init__()

        self.Par = Par
        self.window_size = Par["m"]
        self.optimizer = tf.keras.optimizers.Adam(learning_rate=1e-4)

        self.encoder = GRU(
            32,
            return_sequences=True,
            return_state=True
        )

        self.decoder = GRU(
            32,
            return_sequences=True,
            return_state=True
        )

        self.dense = Dense(self.window_size)

    def call(self, X):
        encoder_sequence, encoder_state = self.encoder(X)

        decoder_output, _ = self.decoder(
            encoder_sequence,
            initial_state=encoder_state
        )

        decoder_output = tf.reshape(
            decoder_output,
            [tf.shape(decoder_output)[0], -1]
        )

        y_pred = self.dense(decoder_output)

        return tf.reshape(y_pred, [-1, self.window_size, 1])


seq_model = Seq2Seq_GRU_Model(Par)

seq_history = train_sequence_model(
    seq_model,
    "Seq2Seq-GRU",
    X_train_seq,
    y_train_seq,
    X_test_seq,
    y_test_seq
)

seq_pred = seq_model(X_test_seq).numpy()

seq_err = relative_l2_error(seq_pred, y_test_seq)

print("Seq2Seq relative L2 error:", seq_err)

plot_convergence(
    seq_history,
    "Seq2Seq-GRU convergence",
    "/content/seq2seq_convergence.png"
)

plot_prediction(
    t_output,
    y_test,
    seq_pred[:, :, 0],
    "Seq2Seq-GRU prediction",
    "/content/seq2seq_prediction.png",
    idx=0
)
```

```python
class DeepONet(Model):
    def __init__(self, latent_dim=100):
        super().__init__()

        self.optimizer = tf.keras.optimizers.Adam(learning_rate=1e-4)

        self.branch = tf.keras.Sequential([
            Dense(128, activation="tanh"),
            Dense(128, activation="tanh"),
            Dense(128, activation="tanh"),
            Dense(latent_dim)
        ])

        self.trunk = tf.keras.Sequential([
            Dense(128, activation="tanh"),
            Dense(128, activation="tanh"),
            Dense(128, activation="tanh"),
            Dense(latent_dim)
        ])

        self.bias = tf.Variable(tf.zeros([1]), trainable=True)

    def call(self, inputs):
        X_func, X_loc = inputs

        b = self.branch(X_func)
        tau = self.trunk(X_loc)

        y = tf.einsum("bk,tk->bt", b, tau)

        return y + self.bias


@tf.function
def train_step_operator(model, X_func, X_loc, y):
    with tf.GradientTape() as tape:
        y_hat = model([X_func, X_loc])
        loss = tf.reduce_mean(tf.square(y_hat - y))

    gradients = tape.gradient(loss, model.trainable_variables)
    model.optimizer.apply_gradients(zip(gradients, model.trainable_variables))

    return loss


def train_operator_model(model, name, X_train, X_loc, y_train, X_test, y_test, n_epochs=3000, batch_size=150):
    history = {"epoch": [], "train_loss": [], "val_loss": []}
    begin_time = time.time()

    for epoch in range(n_epochs + 1):
        for end in np.arange(batch_size, X_train.shape[0] + 1, batch_size):
            start = end - batch_size
            loss = train_step_operator(
                model,
                X_train[start:end],
                X_loc,
                y_train[start:end]
            )

        if epoch % 500 == 0:
            y_hat = model([X_test, X_loc]).numpy()
            val_loss = np.mean((y_hat - y_test) ** 2)

            print(
                f"{name} epoch {epoch}: "
                f"train={loss.numpy():.3e}, "
                f"val={val_loss:.3e}, "
                f"time={int(time.time()-begin_time)}s"
            )

            history["epoch"].append(epoch)
            history["train_loss"].append(loss.numpy())
            history["val_loss"].append(val_loss)

    return history


deeponet = DeepONet(latent_dim=100)

deeponet_history = train_operator_model(
    deeponet,
    "DeepONet",
    X_train_func,
    X_loc,
    y_train_op,
    X_test_func,
    y_test_op
)

deeponet_pred = deeponet([X_test_func, X_loc]).numpy()

deeponet_err = relative_l2_error(deeponet_pred, y_test_op)

print("DeepONet relative L2 error:", deeponet_err)

plot_convergence(
    deeponet_history,
    "DeepONet convergence",
    "/content/deeponet_convergence.png"
)

plot_prediction(
    t_output,
    y_test,
    deeponet_pred,
    "DeepONet prediction",
    "/content/deeponet_prediction.png",
    idx=0
)
```

```python
# ============================================================
# CUSTOM PHYSICS-GUIDED NEURAL OPERATOR — CORRECTED
# ============================================================

class PhysicsGuidedBubbleOperator(tf.keras.Model):
    def __init__(self, latent_dim=128):
        super().__init__()

        self.latent_dim = latent_dim

        self.optimizer = tf.keras.optimizers.Adam(
            learning_rate=1e-4
        )

        self.branch = tf.keras.Sequential([
            tf.keras.layers.Dense(128, activation="tanh"),
            tf.keras.layers.Dense(128, activation="tanh"),
            tf.keras.layers.Dense(128, activation="tanh"),
            tf.keras.layers.Dense(latent_dim)
        ])

        self.trunk = tf.keras.Sequential([
            tf.keras.layers.Dense(128, activation="tanh"),
            tf.keras.layers.Dense(128, activation="tanh"),
            tf.keras.layers.Dense(128, activation="tanh"),
            tf.keras.layers.Dense(latent_dim)
        ])

        self.correction = tf.keras.Sequential([
            tf.keras.layers.Dense(128, activation="tanh"),
            tf.keras.layers.Dense(128, activation="tanh"),
            tf.keras.layers.Dense(m)
        ])

        self.bias = tf.keras.layers.Dense(
            1,
            use_bias=True
        )

    def call(self, inputs, training=False):
        X_func, X_loc = inputs

        X_func = tf.convert_to_tensor(
            X_func,
            dtype=tf.float32
        )

        X_loc = tf.convert_to_tensor(
            X_loc,
            dtype=tf.float32
        )

        branch_features = self.branch(
            X_func,
            training=training
        )

        trunk_features = self.trunk(
            X_loc,
            training=training
        )

        operator_part = tf.einsum(
            "bk,tk->bt",
            branch_features,
            trunk_features
        )

        correction_part = self.correction(
            X_func,
            training=training
        )

        output = operator_part + 0.1 * correction_part

        return output


def train_step_operator(model, X_func, X_loc, y):
    X_func = tf.convert_to_tensor(
        X_func,
        dtype=tf.float32
    )

    X_loc = tf.convert_to_tensor(
        X_loc,
        dtype=tf.float32
    )

    y = tf.convert_to_tensor(
        y,
        dtype=tf.float32
    )

    with tf.GradientTape() as tape:
        y_hat = model(
            [X_func, X_loc],
            training=True
        )

        loss = tf.reduce_mean(
            tf.square(y_hat - y)
        )

    gradients = tape.gradient(
        loss,
        model.trainable_variables
    )

    model.optimizer.apply_gradients(
        zip(
            gradients,
            model.trainable_variables
        )
    )

    return loss


def train_operator_model(
    model,
    name,
    X_train,
    X_loc,
    y_train,
    X_test,
    y_test,
    n_epochs=3000,
    batch_size=150
):
    history = {
        "epoch": [],
        "train_loss": [],
        "val_loss": []
    }

    begin_time = time.time()

    # Build model once before training.
    _ = model(
        [
            X_train[:1],
            X_loc
        ],
        training=False
    )

    for epoch in range(n_epochs + 1):

        for end in np.arange(
            batch_size,
            X_train.shape[0] + 1,
            batch_size
        ):
            start = end - batch_size

            loss = train_step_operator(
                model,
                X_train[start:end],
                X_loc,
                y_train[start:end]
            )

        if epoch % 500 == 0:
            y_hat = model(
                [
                    X_test,
                    X_loc
                ],
                training=False
            ).numpy()

            val_loss = np.mean(
                (y_hat - y_test) ** 2
            )

            print(
                f"{name} epoch {epoch}: "
                f"train={loss.numpy():.3e}, "
                f"val={val_loss:.3e}, "
                f"time={int(time.time() - begin_time)}s"
            )

            history["epoch"].append(epoch)
            history["train_loss"].append(loss.numpy())
            history["val_loss"].append(val_loss)

    return history


custom_operator = PhysicsGuidedBubbleOperator(
    latent_dim=128
)

custom_history = train_operator_model(
    custom_operator,
    "Custom Physics-Guided Operator",
    X_train_func,
    X_loc,
    y_train_op,
    X_test_func,
    y_test_op
)

custom_pred = custom_operator(
    [
        X_test_func,
        X_loc
    ],
    training=False
).numpy()

custom_err = relative_l2_error(
    custom_pred,
    y_test_op
)

print(
    "Custom operator relative L2 error:",
    custom_err
)

plot_convergence(
    custom_history,
    "Custom physics-guided operator convergence",
    "/content/custom_operator_convergence.png"
)

plot_prediction(
    t_output,
    y_test,
    custom_pred,
    "Custom physics-guided operator prediction",
    "/content/custom_operator_prediction.png",
    idx=0
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import time
from sklearn.preprocessing import StandardScaler
from scipy.interpolate import interp1d

def preprocess(dataset, m, npoints_output, is_test=False):
    p = dataset['del_p']
    t = dataset['t']
    r = dataset['R']

    P = interp1d(t, p, kind='cubic')
    R = interp1d(t, r, kind='cubic')

    t_min = 0
    t_max = 5 * 10**-5

    X_func = P(np.linspace(t_min, t_max, m))  # [1500, m]
    # X_loc  = np.linspace(t_min, t_max, npoints_output)[:, None] #[npoints_output,1]
    y = R(np.linspace(t_min, t_max, m))  # [1500, npoints_output]

    if is_test:
        X_func = X_func[:50]
        y = y[:50]

    return X_func, y

def normalize(X_func, y, Par):
    X_func = (X_func - Par['p_mean'])/Par['p_std']
    y = (y - Par['r_mean'])/Par['r_std']

    return X_func.astype(np.float32), y.astype(np.float32)
```

```python
data_train = np.load('/content/res_train.npz')
data_test = np.load('/content/res_test.npz')
m = 20
npoints_output = 500
X_train, y_train = preprocess(data_train, m, npoints_output)
X_test, y_test = preprocess(data_test, m, npoints_output, is_test=True)
```

```python
Par = {}
Par['p_mean'] = np.mean(X_train)
Par['p_std']  = np.std(X_train)

Par['r_mean'] = np.mean(y_train)
Par['r_std']  = np.std(y_train)
Par['m'] = m
Par['address'] = '/content/'
X_train, y_train = normalize(X_train, y_train, Par)
X_test, y_test    = normalize(X_test, y_test, Par)
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import Model, Sequential
from tensorflow.keras.layers import Dense,LSTM

class LSTM_Model(Model):

    def __init__(self, Par):
        super(LSTM_Model, self).__init__()
        np.random.seed(23)
        tf.random.set_seed(23)

        self.Par = Par

        self.index_list = []
        self.train_loss_list = []
        self.val_loss_list = []

        self.lr=10**-4
        self.optimizer = tf.keras.optimizers.Adam(learning_rate=self.lr)

        self.model = Sequential()
        self.model.add(LSTM(50))
        self.model.add(Dense(Par['m']))

    @tf.function()
    def call(self, X):
        X = tf.convert_to_tensor(X)
        pred = self.model(X)
        return tf.reshape(pred, X.shape)

    @tf.function()
    def loss(self, y_pred, y_train):

        #-------------------------------------------------------------#
        #Total Loss
        train_loss = tf.reduce_mean(tf.square(y_pred - y_train))
        #-------------------------------------------------------------#

        return([train_loss])



def train(model, X, y):

    X = tf.convert_to_tensor(
        X,
        dtype=tf.float32
    )

    y = tf.convert_to_tensor(
        y,
        dtype=tf.float32
    )

    with tf.GradientTape() as tape:

        y_hat = model(
            X,
            training=True
        )

        loss = tf.reduce_mean(
            tf.square(y_hat - y)
        )

    gradients = tape.gradient(
        loss,
        model.trainable_variables
    )

    grads_and_vars = [
        (g, v)
        for g, v in zip(
            gradients,
            model.trainable_variables
        )
        if g is not None
    ]

    if len(grads_and_vars) == 0:
        raise ValueError(
            "No gradients found. Check that the model call uses trainable layers and returns a tensor connected to the model variables."
        )

    model.optimizer.apply_gradients(
        grads_and_vars
    )

    return loss
```

```python
model = LSTM_Model(Par)

_ = model(
    tf.convert_to_tensor(
        X_train[:1],
        dtype=tf.float32
    ),
    training=False
)

n_epochs = 10000
batch_size = 150

print("Model Training Begins")
begin_time = time.time()
for i in range(n_epochs+1):
    for end in np.arange(batch_size, X_train.shape[0]+1, batch_size):
        start = end - batch_size
        loss = train(model, X_train[start:end], y_train[start:end])

    if i % 1000 == 0:
        model.save_weights(
    Par['address']
    + "/model_"
    + str(i)
    + ".weights.h5"
)

        train_loss = loss.numpy()

        y_hat = model(X_test)

        val_loss = np.mean((y_hat - y_test)**2)

        print("epoch:" + str(i) + ", Train Loss:" + "{:.3e}".format(train_loss) + ", Val Loss:" + "{:.3e}".format(
            val_loss) + ", elapsed time: " + str(int(time.time()-begin_time)) + "s")

        model.index_list.append(i)
        model.train_loss_list.append(train_loss)
        model.val_loss_list.append(val_loss)

```
